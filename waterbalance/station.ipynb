```python
# get data from this endpoint : https://api-v3.thaiwater.net/api/v1/thaiwater30/public/waterlevel_load
# using requests library
import requests

def get_data():
    url = "https://api-v3.thaiwater.net/api/v1/thaiwater30/public/waterlevel_load"
    response = requests.get(url)
    if response.status_code == 200:
        data = response.json()
        return data
    else:
        print("Error:", response.status_code)
        return None
    
def save_data_to_geojson(data, filename="waterlevel_data.geojson"):
    """ this is a sample of data{
        "id": 1144479462,
        "waterlevel_datetime": "2025-10-04 11:00",
        "waterlevel_m": null,
        "waterlevel_msl": "42.57",
        "waterlevel_msl_previous": "42.56",
        "flow_rate": null,
        "discharge": "421.50",
        "storage_percent": "144.99",
        "sort_order": null,
        "station_type": "tele_waterlevel",
        "situation_level": 5,
        "agency": {
          "id": 12,
          "agency_name": {
            "th": "กรมชลประทาน ",
            "en": "Royal Irrigation Department",
            "jp": ""
          },
          "agency_shortname": {
            "th": "ชป.",
            "en": "RID",
            "jp": ""
          }
        },
        "basin": {
          "id": 8,
          "basin_code": 8,
          "basin_name": {
            "th": "ลุ่มน้ำยม",
            "en": "Yom"
          }
        },
        "station": {
          "id": 2941,
          "tele_station_name": {
            "th": "บางระกำ"
          },
          "tele_station_lat": 16.762039,
          "tele_station_long": 100.121017,
          "tele_station_oldcode": "Y.16",
          "tele_station_type": "tele_waterlevel",
          "left_bank": 70.525999,
          "right_bank": 71.552999,
          "min_bank": 38.93,
          "ground_level": 30.84,
          "offset": 31.629999,
          "sub_basin_id": 215,
          "agency_id": 12,
          "geocode_id": 6174,
          "hydro_id": 2,
          "qmax": 230,
          "is_key_station": true,
          "warning_level_m": null,
          "critical_level_m": null,
          "critical_level_msl": null
        },
        "geocode": {
          "area_code": "3",
          "area_name": {
            "th": "ภาคเหนือ",
            "en": "North"
          },
          "amphoe_code": "04",
          "amphoe_name": {
            "th": "บางระกำ",
            "en": "Bang Rakam District"
          },
          "tumbon_code": "01",
          "tumbon_name": {
            "th": "บางระกำ",
            "en": "Bang Rakam"
          },
          "province_code": "65",
          "province_name": {
            "th": "พิษณุโลก",
            "en": "Phitsanulok"
          }
        },
        "diff_wl_bank": "3.64",
        "diff_wl_bank_text": "ล้นตลิ่ง (ม.)",
        "river_gid": 92709,
        "river_name": "แม่น้ำยม"
      },"""
    features = []
    for item in data:
        feature = {
            "type": "Feature",
            "geometry": {
                "type": "Point",
                "coordinates": [item["station"]["tele_station_long"], item["station"]["tele_station_lat"]]
            },
            "properties": {
                "id": item["id"],
                "station_name": item["station"]["tele_station_name"]["th"],
                "tele_station_oldcode": item["station"]["tele_station_oldcode"],
                "tele_station_name": item["station"]["tele_station_name"].get("th", None),
                "waterlevel_datetime": item["waterlevel_datetime"],
                "waterlevel_msl": item["waterlevel_msl"],
                "discharge": item["discharge"],
                "storage_percent": item["storage_percent"],
                "agency": item["agency"]["agency_shortname"]["th"],
                "basin": item["basin"]["basin_name"]["th"],
                "province": item["geocode"]["province_name"]["th"],
                "amphoe": item["geocode"]["amphoe_name"]["th"],
                "tumbon": item["geocode"]["tumbon_name"]["th"],
                "river_name": item.get("river_name", None)
            }
        }
        features.append(feature)
    
    geojson = {
        "type": "FeatureCollection",
        "features": features
    }
    
    with open(filename, 'w', encoding='utf-8') as f:
        import json
        json.dump(geojson, f, ensure_ascii=False, indent=4)
    print(f"Data saved to {filename}")

if __name__ == "__main__":
    data = get_data()
    if data:
        save_data_to_geojson(data['waterlevel_data']['data'])
```
